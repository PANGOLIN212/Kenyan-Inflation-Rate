```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv(r"C:\Users\hp\Downloads\kenya_inflation_2000_2025.csv")

# Line plot
fig, ax = plt.subplots(figsize=(9, 4.8))
ax.axhspan(2.5, 7.5, color="green", alpha=0.15, label="CBK target band")
ax.plot(df["Year"], df["Inflation_%"], marker="o", label="Inflation rate")
ax.set_title("Kenya: Annual Consumer Price Inflation, 2000–2025")
ax.set_xlabel("Year")
ax.set_ylabel("Inflation rate (%)")
ax.legend()
fig.savefig("line.png", dpi=200)

# Colour-coded bar chart
colors = ["red" if v > 7.5 else "green" if v >= 2.5 else "orange"
          for v in df["Inflation_%"]]
fig, ax = plt.subplots(figsize=(9, 4.8))
ax.bar(df["Year"], df["Inflation_%"], color=colors)
ax.axhline(7.5, ls="--", color="black")
ax.axhline(2.5, ls=":", color="black")
ax.set_title("Kenya: Annual Inflation vs. CBK Target Band")
ax.set_xlabel("Year")
ax.set_ylabel("Inflation rate (%)")
fig.savefig("bar.png", dpi=200)
```
